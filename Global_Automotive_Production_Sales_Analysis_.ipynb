- **Unit Price Outliers:** If you sell both low-cost items (e.g., $5 pack of pens) and high-cost items (e.g., $1,200 laptop), a boxplot or statistical test will flag the laptop as a massive outlier. However, this is correct data, not an error.
- **Revenue Outliers:** Revenue is calculated as Unit Price*Quantity Sold. A single large corporate order or the sale of a few high-ticket items will naturally create huge revenue spikes that look like outliers.

Instead of deleting outliers and losing valuable data, **we will split the dataset into Standard Sales and Fleet Sales based on the units_count.** This separates large-volume commercial orders from smaller everyday purchases, ensuring that high-volume fleet operations don't distort our standard sales analysis.

<p align="left">
  <font size="3" color="Navy"><b><u>Shipping Time/Sales_Velocity:</b></u>
  </font>
  </p>
  
  - **How many days does it take between a car's production date and its actual sales date across different regions?**

```python
data1['Days_to_sell'] = data1['Days_to_sell'].dt.days
```

```python
# How many days does it take between a car's production date and its actual sales date across different regions?

region_days= data1.groupby(['Sales_Region'])['Days_to_sell'].mean().reset_index()
region_days
```

```python
plt.figure(figsize=(8, 6))
sns.set_theme(style="darkgrid",palette='deep')

plt.hlines(y=region_days['Sales_Region'], xmin=31.5, xmax=region_days['Days_to_sell'], color='black', linewidth=2)
plt.plot(region_days['Days_to_sell'], region_days['Sales_Region'], "o", color='black', markersize=8)
for idx, row in region_days.iterrows():
    plt.text(x=row['Days_to_sell'] + 0.05,y=row['Sales_Region'],s=f"{row['Days_to_sell']:.2f}",va='center', ha='left',fontsize=10,weight='bold',color='Black')
plt.title('Sales Velocity: Average Days to Sell by Region', fontsize=14, pad=15)
plt.xlabel('Average Days to Sell')
plt.ylabel('Sales Region')
plt.tight_layout()
plt.show()
```

<p align="left">
  <font size="3" color="Orange"><b><u>Interpretation:</b></u></font></p>

Almost all regions are taken avg 32 days.

  - **Best Performing Region:** Latin America (32.03 days). In automotive sales, a lower number of days to sell indicates higher demand and efficient inventory turnover.

  - **Worst Performing Region:** North America (32.53 days). Vehicles sit on lots the longest here compared to other regions.

  - **Statistical Variance:** The difference between the highest and lowest regions is just 0.50 days (approximately 12 hours). This suggests that global automotive production sales velocity is highly uniform across major markets.

  - **Traditional vs. B2B Channels:** Fleet Enterprise Sales (35,022.30 USD ),Dealership Network (34,913.50 USD) slightly, but both remain very close in overall profit performance.

  - **Margin Disparity:** The difference between the highest-performing channel (Online) and the lowest (Dealership) is only 1,210.20 USD. This indicates relatively stable, standardized pricing and cost structures across all distribution models.

**Taking Vehicle types for forecasting:**

100 \times (1 + 0.15) = 115

```python
# Forecast the next 3 months
FORECAST_MONTHS = 3
SAFETY_STOCK_RATE = 0.15

forecast_results = []
forecast_series = {}
```

```python
monthly_units = forecast_data.groupby(pd.Grouper(key='Sales_Date', freq='MS'))['Units_Count'].sum()
monthly_units
```

```python
# Historical monthly units sold
monthly_units = (
        forecast_data.set_index('Sales_Date')['Units_Count']
        .resample('MS').sum()
    )
monthly_units
```

```python

    # Keep all months continuous; absent months are treated as zero sales
monthly_units = monthly_units.asfreq('MS', fill_value=0).astype(float)
monthly_units
```

```python
forecast_data['Vehicle_Category']
```

```python


for category in unique_categories:
 if len(monthly_units) < 6:
    print("Not enough monthly history to calculate a 6-month average.")

else:
    # Force every category to use the same simple rule directly
  average_units = monthly_units.tail(6).mean()
  method_name = '6-month moving average'


future_dates = pd.date_range(
        start=monthly_units.index[-1] + pd.offsets.MonthBegin(1),
        periods=FORECAST_MONTHS,
        freq='MS'
    )
predicted_units = pd.Series(
        data=[average_units] * FORECAST_MONTHS,
        index=future_dates
    )
```

<font size="4" color="green"><b><u>Prescriptive Analytics — Optimizing Truck Discount Decisions</b></u></font></p>

**Objective:** Identify the observed truck discount range that generates the highest total revenue while meeting a minimum profit-margin requirement of 8%.

**Business purpose:** Support discount decisions while balancing revenue generation and profitability.

```python
MIN_PROFIT_MARGIN_PCT = 8.0

truck_data = data1[data1['Vehicle_Category'].astype(str).str.strip().str.lower()== 'truck'].copy()
truck_data
```

```python
if truck_data.empty:
    print("No valid Truck records found.")
else:
    # Convert discount to percentage
    if truck_data['Discount'].max() <= 1:
        truck_data['Discount_Pct'] = truck_data['Discount'] * 100
    else:
        truck_data['Discount_Pct'] = truck_data['Discount']

    # levels
    bins = [0, 5, 10, 15, 20, 25, 30, np.inf]

    labels = [
        '0–5%', '5–10%', '10–15%', '15–20%',
        '20–25%', '25–30%', '30%+'
    ]
```

```python
truck_data['Discount_Band'] = pd.cut(truck_data['Discount_Pct'],
        bins=bins,
        labels=labels,
        right=False,
        include_lowest=True
    )
```

```python
truck_data['Discount_Band'].isnull().sum()
```

```python
# Summarize historical performance
summary = (truck_data.groupby('Discount_Band', observed=True).agg(Transactions=('Units_Count', 'size'),Units_Sold=('Units_Count', 'sum'),
                                                                                                   Total_Revenue_USD=('Total_Revenue_USD', 'sum'),
                                                                                                   Total_Recorded_Profit_USD=('Net_Profit_Margin_USD', 'sum')).reset_index())
```

```python
# Calculate profit margin

summary['Profit_Margin_Pct'] = np.where(
    summary['Total_Revenue_USD'] > 0,
    summary['Total_Recorded_Profit_USD'] / summary['Total_Revenue_USD'] * 100,
    np.nan
)
```

```python
summary = summary.round({
        'Total_Revenue_USD': 2,
        'Total_Recorded_Profit_USD': 2,
        'Profit_Margin_Pct': 2
    })


display(summary)
```

```python
#  Apply the 8% margin constraint
eligible = summary[(summary['Transactions'] > 0) &
                  (summary['Total_Revenue_USD'] > 0) &
                  (summary['Profit_Margin_Pct'] >= MIN_PROFIT_MARGIN_PCT)].copy()

MIN_PROFIT_MARGIN_PCT
```

```python
# eligible band with highest revenue

best_band = eligible.sort_values('Total_Revenue_USD',ascending=False).iloc[0]
best_band
```

```python
fig, ax1 = plt.subplots(figsize=(11, 5))
ax1.bar(summary['Discount_Band'].astype(str),summary['Total_Revenue_USD'])
ax1.set_xlabel('Truck Discount Band')
ax1.set_ylabel('Total Revenue (USD)')
ax1.tick_params(axis='x', rotation=35)
ax2 = ax1.twinx()
ax2.plot(summary['Discount_Band'].astype(str),summary['Profit_Margin_Pct'],marker='o',linestyle='--',color='darkred',label='Profit Margin')
ax2.axhline(MIN_PROFIT_MARGIN_PCT,linestyle=':',color='green',label='8% Minimum Margin')
ax2.set_ylabel('Recorded Profit Margin (%)')
ax2.legend(loc='best')
plt.title('Truck Discount Optimization')
plt.tight_layout()
plt.show()
```

<font size="3" color="Orange"><b><u>Interpredation:</b></u></font></p>

- **0–5% Discount is the winner:** It makes the absolute most money (highest blue bar) and keeps the highest profit for every dollar sold (highest red dot).

- **High discounts hurt performance:** When you give big discounts (10–15%), your sales drop drastically, and you make less profit on those sales.

- **Everything is safe:** All your categories are well above the green line, meaning you are safely making money across the board and not losing cash on any of these discounts.

**The Bottom Line:** Keep discounts low (0–5%). It drives the highest sales volume and keeps your business the most profitable.
